# Linear Regression



```python
import numpy as np
import matplotlib.pyplot as plt

## training set
# x is size in 1000 sq ft
x_train = np.array([1.0, 2.0])
# y is price in 1000s of dollars
y_train = np.array([300.0, 500.0])

print(f"Number of training examples is: {len(x_train)}")

# Plot the data points
plt.scatter(x_train, y_train, marker='x', c='r')
# Set the title
plt.title("Housing Prices")
# Set the y-axis label
plt.ylabel('Price (in 1000s of dollars)')
# Set the x-axis label
plt.xlabel('Size (1000 sqft)')
plt.show()
```

```python
# Model function
w = 200
b = 100
print(f"w: {w}, b: {b}")

# x^(0) -> f_wb = w * x[0] + b
# x^(1) -> f_wb = w * x[1] + b
def compute_model_output(x, w, b):
    m = len(x)
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = w * x[i] + b

    return f_wb

out_wb = compute_model_output(x_train, w, b)

# Plot our model prediction
plt.plot(x_train, out_wb, c='b',label='Our Prediction')

# Plot the data points
plt.scatter(x_train, y_train, marker='x', c='r',label='Actual Values')

# Set the title
plt.title("Housing Prices")
# Set the y-axis label
plt.ylabel('Price (in 1000s of dollars)')
# Set the x-axis label
plt.xlabel('Size (1000 sqft)')
plt.legend()
plt.show()
```

```python
cost = w * (1.2) + b
print(f"Predicted cost: {cost * 1000}")
```

# Gradient Descent
- Minimize cost function for a model
- Minimize J(w,w2,w3..., b)

1. Start with w = 0, b = 0
2. Keep changing w,b to reduce J(w,b) "spin 360, move step forward that reduces cost"
3. Until we settle at or near minimum (possible to have more than one minumum!)
